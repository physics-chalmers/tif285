Inspecting these plots we learn that:
- The amplitude of the output signal scales as $\sqrt{N_h} \sigma_\mathrm{out}$, where $N_h$ is the number of hidden units and $\sigma_\mathrm{out}$ the initialization width of the output weights and bias.
- The length scale over which we see significant changes in the outsput signal scales as $\sigma_\mathrm{bias}/\sigma_\mathrm{weight}$, where these are the magnitudes of the hidden units biases and weights, respectively.
- The shortest horizontal length scale is of order $1/\sigma_\mathrm{weight}$.
